# Türkçe Erkek Sesi — Fahrettin
Reels Stüdyo için ücretsiz erkek sesi WAV üretimi. Google Colab'da sırayla hücreleri çalıştır. Bu yöntem uygulama içinde doğrudan ses üretmez; WAV dosyasını indirip editöre eklersin. Modelin kullanım koşullarını ayrıca kontrol et.

```python
!pip -q install piper-tts
import subprocess
print('Piper kuruldu')
```

```python
#@title Erkek sesiyle okunacak Türkçe metin
metin = 'Merhaba, bugün yeni bir hikâye anlatacağım.' #@param {type:'string'}
if not metin.strip(): raise ValueError('Metin boş olamaz')
if len(metin)>2500: raise ValueError('Metin en fazla 2500 karakter olabilir')
cmd=['piper','--model','tr_TR-fahrettin-medium','--output_file','erkek_sesi.wav']
subprocess.run(cmd,input=metin,text=True,check=True)
from IPython.display import display,Audio
display(Audio('erkek_sesi.wav'))
from google.colab import files
files.download('erkek_sesi.wav')
```
